```python
#!/usr/bin/env python3
# -*- coding: utf-8 -*-

'''
Created on 2024-08-16
Last modified on 2024-08-16
@author: Juan Enrique López
@description: Tras la ejecución de get_rules_and_classify_by_ttp.ipynb se generan tantas carpetas como dominios Mitre, fuentes y ttps son analizadas.Mediante este script generamos una única carpeta que contiene las reglas disponibles clasificadas por dominio y id de TTP.
Adicionalmente genera un archivo zip o  7z con todas las carpetas.

'''
```

**Requerimientos**

```python
import os
import shutil
import zipfile
import py7zr
```

```python
def unify_ttps_rules(base_path, new_base_path):
    '''
    Función que nos permite recorrer la ruta original donde se clasificaron las reglas por id de ttp, source y dominio para posteriormente copiarlas a una nueva carpeta donde únicamente estén clasificadas por id de ttp.
    '''
    # Recorremos la path dada para obtener las subcarpetas (sources/fuente de reglas)
    for domain in os.listdir(base_path):
        domain_path = os.path.join(base_path, domain)
        for source in os.listdir(domain_path):
            source_path = os.path.join(domain_path, source)
            
            if os.path.isdir(source_path):
                # Guardamos en una variable el nombre para posteriormente añadirlo al nombre del fichero
                source_name = os.path.basename(source_path)
                # Recorremos la fuente para obtener la lista de ttps generadas y almacenamos en una variable para guardar posteriormente
                for ttp in os.listdir(source_path):
                    ttp_path = os.path.join(source_path, ttp)
                    if os.path.isdir(ttp_path):
                        ttp_name = os.path.basename(ttp_path)
                        # Creamos la carpeta de la ttp correspondiente en la carpeta de unificación
                        new_ttp_path = os.path.join(new_base_path, ttp_name)
                        if not os.path.exists(new_ttp_path):
                            os.makedirs(new_ttp_path)
                        # Copiamos y renombramos los archivos
                        for file_name in os.listdir(ttp_path):
                            file_path = os.path.join(ttp_path, file_name)
                            if os.path.isfile(file_path):
                                new_file_name = f"[{source_name}]_{file_name}"
                                new_file_path = os.path.join(new_ttp_path, new_file_name)
                                # Copiamos el archivo a la nueva ubicación
                                shutil.copy(file_path, new_file_path)
                                print(f"Archivo copiado: {file_path} -> {new_file_path}")
    print('Proceso finalizado con éxito')
```

```python
def create_zip_from_folder(folder_path):
    '''
    Función cuyo cometido es comprimir la carpeta pasada por argumento en formato zip.
    '''
    folder_name = os.path.basename(folder_path.rstrip('/'))
    zip_file_name = f"{folder_name}.zip"
    
    with zipfile.ZipFile(zip_file_name, 'w', zipfile.ZIP_DEFLATED) as zipf:
        for root, dirs, files in os.walk(folder_path):
            for file in files:
                file_path = os.path.join(root, file)
                relative_path = os.path.relpath(file_path, folder_path)
                zipf.write(file_path, relative_path)
    
    print(f"Archivo zip creado: {zip_file_name}")
```

```python
def create_7z_from_folder(folder_path):
    '''
    Función cuyo cometido es comprimir la carpeta pasada por argumento en formato 7z. La compresión 7z generalmente ofrece una mayor tasa de compresión y soporte para más algoritmos avanzados (como LZMA2), mientras que ZIP es más rápido y ampliamente compatible, pero con una tasa de compresión inferior.
    '''
    folder_name = os.path.basename(folder_path.rstrip('/'))
    seven_zip_file_name = f"{folder_name}.7z"
    
    filters = [{
        'id': py7zr.FILTER_LZMA2,
        'preset': 9,  # Nivel de compresión más alto
    }]
    
    # Crear el archivo .7z
    archive = py7zr.SevenZipFile(seven_zip_file_name, 'w', filters=filters)
    try:
        # Añadir la carpeta al archivo .7z
        archive.writeall(folder_path, folder_name)
    finally:
        # Cerrar explícitamente el archivo para liberar los recursos
        archive.close()
    
    print(f"Archivo 7z creado: {seven_zip_file_name}")
```

**Parámetros**

```python
input_path = os.path.join(os.getcwd(),'outputs')
output_path = os.path.join(input_path,'unified_outputs')

compression_type = '7z' # 'zip', '7z'
```

**Ejecución**

```python
unify_ttps_rules(input_path, output_path)

if compression_type == 'zip':
    create_zip_from_folder(output_path)
elif compression_type == '7z':
    create_7z_from_folder(output_path)
else:
    pass
```
