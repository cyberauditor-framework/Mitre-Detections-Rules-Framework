```python
from attackcti import attack_client
import pandas as pd
import datetime
import csv
import os
import json
```

```python
def delete_special_characters(df):
    chars = r'[^\w\s,-.:]'  # Esto incluye cualquier cosa que no sea una palabra o espacio
    chars_url = r"[\[\]\'{}]"
    lb = r'\n'
    cr = r'\r'
    low_bar = r'_'
    colon_sep = r"': '"
    df = df.fillna('N/A')
    for col in df.columns:
        # Convertimos a str todas las columna
        if df[col].dtype == 'O':
            df[col] = df[col].astype(str)
        elif df[col].dtype == 'bool':
            df[col] = df[col].replace({True: 'True', False: 'False'})
        if col != 'external_references':
            # Aplicamos los reemplazos, importante mantener este orden
            df[col] = df[col].str.replace(colon_sep, ' - ', regex=True)
            df[col] = df[col].str.replace(low_bar, ' ', regex=True)
            df[col] = df[col].str.replace(chars, '', regex=True)
            df[col] = df[col].str.replace(lb, ' ', regex=True)
            df[col] = df[col].str.replace(cr, '', regex=True)
        else:
            df[col] = df[col].str.replace(chars_url, '', regex=True)
            df[col] = df[col].str.replace(colon_sep, ' - ', regex=True)
            df[col] = df[col].str.replace(low_bar, ' ', regex=True)
            df[col] = df[col].str.replace(lb, ' ', regex=True)
            df[col] = df[col].str.replace(cr, '', regex=True)
    return df
```

```python
def save_list_as_csv(method, info_as_list):
    now = datetime.datetime.now()
    file_to_save = os.path.join(os.getcwd(), 'outputs',('MITRE_' +method+ '_'+ now.strftime('%d%m%Y_%H%M')+'h.csv'))
    with open(file_to_save, mode='w', newline='', encoding='utf-8') as file:
        writer = csv.writer(file)
        for item in info_as_list:
            writer.writerow([item])
    print('Fichero generado correctamente '+ method)
```

```python
def save_df_as_csv(df, method):
    now = datetime.datetime.now()
    file_to_save = os.path.join(os.getcwd(), 'outputs',('MITRE_df_' +method+ '_'+ now.strftime('%d%m%Y_%H%M')+'h.csv'))
    df.to_csv(file_to_save, sep=';', encoding='utf8', index=False, quoting=csv.QUOTE_NONNUMERIC)
    print('Fichero generado correctamente '+ method)
```

```python
methods_list = ['get_enterprise_tactics','get_techniques_used_by_all_groups','get_data_components','get_software','get_techniques','get_enterprise_techniques','get_tactics','get_data_sources','get_mitigations','get_enterprise_mitigations','get_relationships','get_enterprise_relationships', 'get_groups', 'get_enterprise_groups']
columns_method = {}
for method_name in methods_list:
    if method_name:
        method = getattr(lift, method_name, None)
        if callable(method):
            try:
                
                if create_list_csv:
                    mitre_info = method()
                    save_list_as_csv(method_name, mitre_info)
                elif create_df_csv:
                    mitre_info = method(stix_format=False)
                    mitre_info_df = pd.json_normalize(mitre_info)
                    mitre_info_df = delete_special_characters(mitre_info_df)
                    save_df_as_csv(mitre_info_df, method_name)
                    columns_method.update({method_name: mitre_info_df.columns.to_list()})
            except Exception as e:
                print(f"Error al llamar a {method_name}: {e}")
```
