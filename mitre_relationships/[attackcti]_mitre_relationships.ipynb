```python
def delete_special_characters(df):
    chars = r'[^\w\s,-.:]'  # Esto incluye cualquier cosa que no sea una palabra o espacio
    chars_url = r"[\[\]\'{}]"
    lb = r'\n'
    cr = r'\r'
    low_bar = r'_'
    colon_sep = r"': '"
    df = df.fillna('N/A')
    for col in df.columns:
        # Convertimos a str todas las columna
        if df[col].dtype == 'O':
            df[col] = df[col].astype(str)
        elif df[col].dtype == 'bool':
            df[col] = df[col].replace({True: 'True', False: 'False'})
        if col != 'external_references':
            # Aplicamos los reemplazos, importante mantener este orden
            df[col] = df[col].str.replace(colon_sep, ' - ', regex=True)
            df[col] = df[col].str.replace(low_bar, ' ', regex=True)
            df[col] = df[col].str.replace(chars, '', regex=True)
            df[col] = df[col].str.replace(lb, ' ', regex=True)
            df[col] = df[col].str.replace(cr, '', regex=True)
        else:
            df[col] = df[col].str.replace(chars_url, '', regex=True)
            df[col] = df[col].str.replace(colon_sep, ' - ', regex=True)
            df[col] = df[col].str.replace(low_bar, ' ', regex=True)
            df[col] = df[col].str.replace(lb, ' ', regex=True)
            df[col] = df[col].str.replace(cr, '', regex=True)
    return df
```

```python
# Buscamos en la columna de descripción o informativa
def find_techniques(texto, techniques_enterprise_list):
    if isinstance(texto, str):
        found = []
        for item in techniques_enterprise_list:
            if item in texto:
                found.append(item)
        return ', '.join(found)
    else:
        return ''
```

```python
def getListOfFilesSub(dirName):
    # Crea lista de ficheros y subdirectorios
    listOfFile = os.listdir(dirName)
    allFiles = list()
    #Recorre los directorios y subdirectorios y genera una lista con los ficheros encontrados en estos
    for entry in listOfFile:
        fullPath = os.path.join(dirName, entry)
        if os.path.isdir(fullPath):
            allFiles = allFiles + getListOfFilesSub(fullPath)
        else:
            allFiles.append(fullPath)
    #Devuelve la lista con los ficheros
    return allFiles
```

```python
# Función para quedarnos unicamente con items unicos al agregar
def unique_list(series):
    return list(set(series))
```

```python
def cp_unique_ttp_with_rules(path):
    # Conjunto para almacenar nombres únicos de carpetas
    unique_folders = set()
    
    # Recorrer el directorio y sus subdirectorios
    for root, dirs, files in os.walk(path):
        for dir_name in dirs:
            # Verificar si el nombre de la carpeta contiene 'T0' o 'T1'
            if 'T0' in dir_name or 'T1' in dir_name:
                # Añadir solo el nombre de la carpeta al conjunto (esto elimina duplicados automáticamente)
                unique_folders.add(os.path.basename(dir_name))
    
    # Convertir el conjunto a una lista para el resultado final
    filtered_folders = list(unique_folders)
    filtered_folders = sorted(filtered_folders, key=len, reverse=True)
    filtered_folders.remove('T0000')
    
    return filtered_folders
```

```python
def cp_ttps():
    # ttps_outputs_path = r'C:\Users\jelopez\Documents\CyberProof\python\develop\download_ttps\saved'
    ttps_outputs_path = r'C:\Users\jelopez\Documents\CyberProof\python\develop\get_rules_and_classify_by_ttp\outputs\enterprise'
    cp_ttp_available = cp_unique_ttp_with_rules(ttps_outputs_path)
    return cp_ttp_available
```

```python
# El campo data_sources viene con un formato que no nos sirve para asociar el id y la url, por lo que lo formateamos dando lugar a 2 campos nuevos 
def split_lists(lst):
    if lst is None:
        return [], []
    before_colon = [item.split(': ')[0] for item in lst if item]
    after_colon = [item.split(': ')[1] for item in lst if item]
    return before_colon, after_colon
```

Solución: Tras probar directamente a stix y otras opciones he encontrado el método get_techniques_used_by_software(), al cual si se le pasa el objeto stix2.v20.sdo.Tool correspondiente como unico software del metodo get_software() como parametro este devolverá la tabla con las técnicas asociadas. Por lo tanto, mediante un bucle iremos recorriendo la lista de software (794) y se iran añadiendo a un df donde ademas de las técnicas incorporaremos la información relativa al SW.

```python
results_df = pd.DataFrame(columns=['type', 'id', 'url', 'matrix', 'created', 'modified', 'created_by_ref',
       'modified_by_ref', 'external_references', 'object_marking_refs',
       'mitre_version', 'mitre_attack_spec_version', 'mitre_deprecated',
       'technique', 'technique_id', 'technique_description', 'tactic',
       'technique_detection', 'platform', 'data_sources', 'defense_bypassed',
       'permissions_required', 'effective_permissions', 'system_requirements',
       'network_requirements', 'remote_support', 'contributors', 'tactic_type',
       'impact_type', 'is_subtechnique', 'software_id', 'software',
       'software_aliases', 'software_description', 'software_type',
       'url_software', 'mitre_domains'])
```

```python
# En esta parte del código obtenemos la lista de software y uno a uno se lo pasamos al método get_techniques_used_by_software que nos devolverá la tabla con las técnicas utilizadas por cada SW.
# Importante: Este código es lento, aproximadamente tarda 1,5 minutos por petición lo que hace que para los 795 items tarde aproximadamente 20 horas.
# Si se tiene que reejecutar esta parte tambien es posible que falle alguna conexión, en tal caso en lugar de empezar de 0 simplemente no ejecutar la celda anterior y modificar range(len(software)) por range(row, len(software))

software = lift.get_software()

for row in range(len(software)):

    info_row_techniques_software =  lift.get_techniques_used_by_software(software[row], stix_format=False)
    info_row_techniques_software_df = pd.json_normalize(info_row_techniques_software)


    # Recogemos info de software y la añadimos al df
    info_row_techniques_software_df['software_id'] = software[row].external_references[0].external_id

    info_row_techniques_software_df['software'] = software[row].name

    aliases = getattr(software[row], 'x_mitre_aliases', [])
    info_row_techniques_software_df['software_aliases'] = [aliases] * len(info_row_techniques_software_df)

    info_row_techniques_software_df['software_description'] = software[row].description

    info_row_techniques_software_df['software_type'] = software[row].type

    info_row_techniques_software_df['url_software'] = software[row].external_references[0].url
    
    domains = getattr(software[row], 'x_mitre_domains', [])
    info_row_techniques_software_df['mitre_domains'] = [domains] * len(info_row_techniques_software_df)

    print(f"Finalizado {row}: {software[row].name}")


    results_df = pd.concat([results_df,info_row_techniques_software_df])
# Filtramos técnicas deprecadas
results_df = results_df[(results_df['mitre_deprecated'] != 'True')]
# Filtramos SW enterprise
results_df = results_df[results_df['mitre_domains'].apply(lambda x: 'enterprise-attack' in x)]

results_df['url_technique'] = results_df['external_references'].apply(lambda x: x[0]['url'] if x else None)

if save_to_file:
    save_df_as_csv(results_df, '[MITRE]_n2n_main_techniques_software', 'inter_software_techniques')
```

```python
# En este bucle vamos a ir recorriendo cada uno de los grupo y retornando todas los sotfware utilizados por estos. Iremos uniendo toda la información en el df anteriormente creado vacío.
for row in range(row, len(get_groups_df)):
    temp_software_by_group = lift.get_software_used_by_group(get_groups_df.iloc[row], stix_format=False)
    temp_software_by_group_df = pd.json_normalize(temp_software_by_group)

    temp_software_by_group_df['group_id'] = get_groups_df.iloc[row].group_id
    temp_software_by_group_df['group'] = get_groups_df.iloc[row].group
    temp_software_by_group_df['group_description'] = get_groups_df.iloc[row].group_description

    aliases = get_groups_df.iloc[row].group_aliases
    temp_software_by_group_df['group_aliases'] = [aliases] * len(temp_software_by_group_df)

    domains = get_groups_df.iloc[row].matrix
    temp_software_by_group_df['matrix'] = [domains] * len(temp_software_by_group_df)

    temp_software_by_group_df['url_group'] = get_groups_df.iloc[row].url
    temp_software_by_group_df['type'] = get_groups_df.iloc[row].type

    results_df = pd.concat([results_df, temp_software_by_group_df])

    print(f"Finalizado {row}: {get_groups_df.iloc[row].group}")

# Eliminamos duplicados, no debería darse ninguno pero en caso de ejecutar por lotes (cuando se pierde la conexión) es posible que por error humano dupliquemos alguna fila
results_df = results_df.drop_duplicates(subset=['group_id', 'software_id'], keep='last')
# Renombramos la columna url
results_df = results_df.rename(columns={'url': 'url_software'})
```

### **BONUS. LISTADO DE REGLAS DE DETECCIÓN POR TÉCNICA (TTP ID)**

**Importante:** Debe haberse ejecutado previamente el codigo de obtención de las reglas de detección y clasificación por TTP

```python
def get_rules_from_path(path):
    # Listas para almacenar los resultados
    sources = []
    ttps = []
    file_names = []
    file_extension = []

    # Recorremos la path accediendo a cada directorio para poder añadirlo a una lista
    for root, dirs, files in os.walk(path):
        for file in files:
            # Extensión del archivo
            ext = os.path.splitext(file)[1].lower()
            if ext in ['.csv', '.yaml','.yml', '.toml', '.md', '.txt', '.yar']:
                # Última y penúltima subcarpeta
                path_split = root.split(os.sep)
                ttp = path_split[-1]
                source = path_split[-2] if len(path_split) > 1 else ''
                # Añadimos elementos a las listas
                sources.append(source)
                ttps.append(ttp)
                file_names.append(file)
                file_extension.append(ext)

    df = pd.DataFrame({'source': sources, 'TTP': ttps, 'rule': file_names, 'file_extension': file_extension})

    return df
```

```python
technique_matrix = '' # Dejamos preparado para el futuro cercano la posibilidad de lectura de tecnicas por matriz
path = os.path.join(os.path.dirname(os.getcwd()), 'download_ttps\saved', technique_matrix)
ttp_rules_df = get_rules_from_path(path)
if save_to_file:
    save_df_as_csv(ttp_rules_df, '[CP-attactcti]_ttp_rules')
```
